```python
import httpx2
#import pandas as pd
#from pprint import pprint
#import json
from bs4 import BeautifulSoup
```

```python
base_url = "https://openlibrary.org/search/authors.json?"
params = {"author": "John Locke", "fields": "author_key, author_name", "limit": "1"}
get_author_key = httpx2.get(base_url, params=params)
get_author_key.json()
```

```python
author_docs = get_author_key.json()["docs"][0]
author_key = author_docs["author_key"][0]
author_name = author_docs["author_name"][0]

print(author_name, author_key) #Aeschylus OL74434A, Esquilo OL3203166A
author_docs
```

```python
if params["author"] not in author_name:
    for author in author_docs["author_name"]:
        if params["author"] in author: 
            author_name = author
            author_key_index = author_docs["author_name"].index(author) 
            author_key = author_docs["author_key"][author_key_index] 

print(author_name, author_key) 
```
